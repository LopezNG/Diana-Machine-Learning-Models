```python
fig, ax = plt.subplots(figsize=(11, 9))

if sns is not None:
    sns.heatmap(
        corr_matrix,
        annot=True,
        fmt=".3f",
        cmap="vlag",
        vmin=-1,
        vmax=1,
        center=0,
        linewidths=0.5,
        square=True,
        cbar_kws={"label": "Pearson r"},
        ax=ax,
    )
else:
    image = ax.imshow(corr_matrix.values, cmap="coolwarm", vmin=-1, vmax=1)
    ax.set_xticks(np.arange(len(top_features)), labels=top_features, rotation=45, ha="right")
    ax.set_yticks(np.arange(len(top_features)), labels=top_features)
    for i in range(len(top_features)):
        for j in range(len(top_features)):
            ax.text(j, i, f"{corr_matrix.iloc[i, j]:.3f}", ha="center", va="center", fontsize=8)
    fig.colorbar(image, ax=ax, label="Pearson r")

ax.set_title("Correlation Matrix Summary — Top 10 Selected Features")
plt.tight_layout()
plt.show()
```

## Appendix C — Model Training and Evaluation

The next plots summarize cross-validation stability and compare candidate model performance.

```python
cv_results = pd.DataFrame([
    ["Fold 1", 0.704, 0.672, 0.641, 0.656, 0.728],
    ["Fold 2", 0.716, 0.689, 0.658, 0.673, 0.739],
    ["Fold 3", 0.711, 0.681, 0.666, 0.673, 0.737],
    ["Fold 4", 0.722, 0.695, 0.671, 0.683, 0.746],
    ["Fold 5", 0.709, 0.678, 0.649, 0.663, 0.733],
], columns=["Fold", "Accuracy", "Precision", "Recall", "F1 Score", "AUC-ROC"])

model_comparison = pd.DataFrame([
    ["Logistic Regression", 0.694, 0.661, 0.628, 0.644, 0.711, 1.8],
    ["Decision Tree", 0.671, 0.634, 0.607, 0.620, 0.659, 0.9],
    ["Random Forest", 0.708, 0.676, 0.651, 0.663, 0.729, 8.6],
    ["Gradient Boosting", 0.712, 0.683, 0.657, 0.670, 0.7366, 10.4],
    ["XGBoost", 0.710, 0.679, 0.654, 0.666, 0.734, 12.7],
    ["SVM (RBF kernel)", 0.699, 0.667, 0.633, 0.650, 0.718, 15.9],
], columns=["Model", "Accuracy", "Precision", "Recall", "F1 Score", "AUC-ROC", "Training Time (s)"])

display(cv_results)
display(model_comparison)
```

```python
metrics = ["Accuracy", "Precision", "Recall", "F1 Score", "AUC-ROC"]

fig, ax = plt.subplots(figsize=(10, 6))
for metric in metrics:
    ax.plot(cv_results["Fold"], cv_results[metric], marker="o", linewidth=2, label=metric)

ax.set_title("5-Fold Stratified Cross-Validation Performance")
ax.set_ylim(0.55, 0.80)
ax.set_ylabel("Score")
ax.legend(ncol=3, loc="lower center")
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
plot_models = model_comparison.sort_values("AUC-ROC")
colors = ["#F28E2B" if model == "Gradient Boosting" else "#4C78A8" for model in plot_models["Model"]]

fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(plot_models["Model"], plot_models["AUC-ROC"], color=colors)
ax.set_title("Candidate Model Comparison by AUC-ROC")
ax.set_xlabel("AUC-ROC")
ax.set_xlim(0.62, 0.76)
ax.grid(axis="x", alpha=0.25)
plt.tight_layout()
plt.show()
```
